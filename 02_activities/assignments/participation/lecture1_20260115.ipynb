### Lecture codes

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)
```

```python
# Scatter plot
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x, y)
```

```python
# Bar plot
fig, ax = plt.subplots(figsize=(5,3))
ax.bar(x, y)
```

```python
# Line plot
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y)
```

```python
# Histogram
fig, ax = plt.subplots(figsize=(5,3))
ax.hist(y, edgecolor='black')
```

```python
# Add axis labels and titles
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

```python
# Define font dictionaries
font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}
```

```python
# Modify labels and titles with font dictionaries
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
```

```python
# Modify the plot
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(
    x,
    y,
    marker='*', # style of individual points
    color='indigo' # color of individual points
)
plt.show()
```

```python
# Further modifications
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,
        y,
        marker='*',
        markersize=12, # size of individual points
        color="#7425b9", # color of line defined by hex code
        linestyle='--', # style of line
        linewidth=2, # thickness of line
        markeredgecolor='#fa9359', # outline color of individual points
        markerfacecolor='#000000') # fill color of individual points

# Add gridlines
ax.grid(axis='y',
        color='blue',
        linewidth=2,
        linestyle='-.')
plt.show()
```
