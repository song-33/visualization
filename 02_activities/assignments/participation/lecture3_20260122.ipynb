```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)
```

```python
fig,ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1)
ax.plot(x,y2)
plt.show()
```

```python
# Add legend
fig,ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label = 'Person 1') # legend for y1
ax.plot(x,y2, label = 'Person 2') # legend for y2
ax.legend(loc='best') #specify legend location
plt.show()
```

```python
# Modify the legend
fig,ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label = 'Person 1')
ax.plot(x,y2, label = 'Person 2')
ax.legend(loc='lower right',
          frameon=True, #add border
          fontsize=12, #change font size
          ncol=2, #number of columns
          shadow=True) #add shadow
plt.show()
```

```python
# Move the legend outside the plot area
fig,ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label = 'Person 1')
ax.plot(x,y2, label = 'Person 2')
ax.legend(loc='upper left', 
          bbox_to_anchor=(1,1)) #attach the upper left corner of the legend box to the top right corner of the axes
plt.show()
```

```python
# Text annotations
fig,ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = 'Person 1')
ax.scatter(x,y2, label = 'Person 2')
ax.legend(loc='lower right')
ax.text(10, 95, 'This value is important!') #(x position, y position, text)
plt.show()
```

```python
# Modify text annotations
fig,ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = 'Person 1')
ax.scatter(x,y2, label = 'Person 2')
ax.legend(loc='lower right')
ax.text(10, 95, 'This value is important!',
        ha='center', #horizontal alignment to the center
        color='red', #text color
        size=20) #font size
plt.show()
```

```python
# Position text annotations using transforms
fig,ax = plt.subplots()
ax.axis([0,10,0,10]) #set axis limits
ax.text(1,5, '. Data:(1,5)', transform=ax.transData)#position annotation in data coordinates
ax.text(0.5,0.1, '. Axes:(0.5,0.1)', transform=ax.transAxes) #position annotation in axes coordinates
ax.text(0.2,0.2, 'Figure:(0.2,0.2)', transform=fig.transFigure) #position annotation in figure coordinates
```

```python
# Annotate with arrows
fig,ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = 'Person 1')
ax.scatter(x,y2, label = 'Person 2')
ax.legend(loc='lower right')
ax.annotate('This is important!',
            xy=(10,95), #point to annotate
            xytext=(20,94), #position of the text
            arrowprops=dict(facecolor='black', arrowstyle='->')) #arrow properties
```

```python
# Modify arrow properties
fig,ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = 'Person 1')
ax.scatter(x,y2, label = 'Person 2')
ax.legend(loc='lower right')
ax.annotate('This is important!',
            xy=(10,95),
            xytext=(20,94),
            arrowprops=dict(color='hotpink', arrowstyle='fancy')) #different arrow styles: '->', '<->', 'fancy', 'simple', 'wedge' etc.
```

```python
# Remove tick marks and labels
fig,ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = 'Person 1')
ax.scatter(x,y2, label = 'Person 2')
ax.legend(loc='lower right')

ax.yaxis.set_major_locator(plt.NullLocator()) #remove y axis ticks, use locator to remove ticks and labels
ax.xaxis.set_major_formatter(plt.NullFormatter()) #remove x axis tick labels, use formatter to remove labels only
plt.show()
```

```python
# Limit tick marks using MaxNLocator
fig,ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = 'Person 1')
ax.scatter(x,y2, label = 'Person 2')
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(3)) #limit to maximum of 3 ticks on x axis
```

```python
# Tick mark intervals using MultipleLocator
fig,ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = 'Person 1')
ax.scatter(x,y2, label = 'Person 2')
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MultipleLocator(5)) #set x axis ticks at multiples of 5
```

```python
# Rotate axis labels
fig,ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = 'Person 1')
ax.scatter(x,y2, label = 'Person 2')
ax.legend(loc='lower right')

plt.xticks(rotation=45, #rotate x axis labels by 45 degrees
           ha='right', #align to the right
           size=12,
           color='purple'
           ) 
```
