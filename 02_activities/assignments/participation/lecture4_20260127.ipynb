```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x1=np.arange(50)
y1=np.random.randint(0,75,50)
x2=np.array(['Luffy', 'Zoro', 'Nami', 'Usopp', 'Sanji'])
y2=np.array([110,180,240,99,220])
```

```python
# Subplots
fig, (ax1, ax2) = plt.subplots(ncols=2, #two separate axes side by side
                               nrows=1,
                               figsize=(7,3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
plt.show()
```

```python
# Alternative way to index axes
fig, ax = plt.subplots(ncols=2, #two separate axes side by side
                               nrows=1,
                               figsize=(7,3))
ax[0].scatter(x1,y1)
ax[1].bar(x2,y2)
plt.show()
```

```python
# Customize subplots
fig, (ax1, ax2) = plt.subplots(ncols=2, #two separate axes side by side
                               nrows=1,
                               figsize=(7,3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
ax1.annotate('This is important', xy=(10,65), xytext=(30,65),
             arrowprops=dict(facecolor='black', arrowstyle='->'))
plt.show()
```

```python
# Mosaic layout, without a grid arrangment
fig,someaxes = plt.subplot_mosaic([['ax1', 'ax3'], #a list of lists defining layout, first row has ax1 and ax3, second row has ax2 and ax3
                                   ['ax2', 'ax3']], 
                                   figsize=(7,4))
someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)
plt.show()
```

```python
# Layouts
fig,someaxes = plt.subplot_mosaic([['ax1', 'ax3'], 
                                   ['ax2', 'ax3']], 
                                   figsize=(7,4))
someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)
someaxes['ax1'].set_xlabel('A Big Label', fontsize=18) #setting xlabel for each axis
someaxes['ax2'].set_xlabel('Another Label', fontsize=18) #adding this title makes the ax1 title disappear
someaxes['ax3'].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize=18) #title is too big to show the full text
plt.show()
```

```python
# Contrained layout
fig,someaxes = plt.subplot_mosaic([['ax1', 'ax3'], 
                                   ['ax2', 'ax3']], 
                                   figsize=(7,4),
                                   layout='constrained') #constrained layout also fits things like legens or colorbars
someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)
someaxes['ax1'].set_xlabel('A Big Label', fontsize=18) 
someaxes['ax2'].set_xlabel('Another Label', fontsize=18) 
someaxes['ax3'].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize=18) 
plt.show()
```

```python
# Multiple viz on one axis
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

fig, ax = plt.subplots(figsize=(7,3))
ax.bar(x, y1, color='indigo')
ax.plot(x, y2, color='red') #bar plot overlay with line plot
```

```python
# Add error info
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize=(7,3))
#ax.plot(x, y2, color='red')
#ax.errorbar(x, y2) #ax.errorbar() automatically draws lines between points
ax.errorbar(x, y2, 
            yerr=y2_sd, 
            color='red', #color of line
            fmt='-', #add error bars to line plot, fmt specifies data marker, if fmt="none" then no line is drawn between points
            ecolor='blue', #color of error bars
            elinewidth=4, #thickness of error bar lines
            capsize=6, #length of error bar caps
            capthick=4, #thickness of error bar caps
            errorevery=2) #only show error bars for every n-th data point instead of all points
plt.show()

### Extra notes on ax.errorbar() ###
# ax.plot() + ax.errorbar(fmt='none') produces the same result as ax.errorbar(fmt='-')
```
